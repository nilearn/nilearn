```python
import numpy as np

from nilearn.plotting import plot_matrix, show

rng =  np.random.default_rng(0)
matrix = rng.normal(size=(10, 10))
vmax = np.max(np.abs(matrix.ravel()))

fig = plot_matrix(matrix, vmax=vmax, vmin=-vmax)

show()
```
