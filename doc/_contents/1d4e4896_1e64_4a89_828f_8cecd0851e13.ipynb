```python
import pandas as pd

from nilearn.plotting import plot_event, show

trial_type = ["c0", "c0", "c0", "c1", "c1", "c1", "c2", "c2", "c2"]
onset = [0, 70, 100, 10, 30, 90, 30, 40, 60]
duration =  [1, 5, 3] * 3
model_event = pd.DataFrame({"onset": onset,
                            "duration": duration,
                            "trial_type": trial_type})

fig = plot_event(model_event)

show()
```
